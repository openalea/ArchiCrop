```python
def generate_tillers(nb_phy, phyllochron, duration, nb_tillers, tiller_delay):
    '''
    Generates a MTG of a plant with main stem and tillers, considering the delay of appearance of tillers and the timing of appearance and elongation of the phytomers.
    Return a dict with the order and the delay of each tiller.
    '''
    ranks = range(1, nb_phy + 1)
    ntop = max(ranks) - np.array(ranks) + 1
    
    tt = 0
    dtt = phyllochron * duration
    
    tiller_points = []
    rank_tillers = []
    
    g = MTG()
    # Add a root vertex for the plant
    vid_plant = g.add_component(g.root, label="Plant", edge_type="/") 
    # Add a plant vertex for the main axis
    vid_axis = g.add_component(vid_plant, label="MainAxis", edge_type="/")
 
    # iterate over the number of phytomers of the main stem
    first = True
    for rank in ranks:
        if first:
            vid = g.add_component(vid_axis)
            first = False
        else:
            vid = g.add_child(vid, edge_type="<")
        g.node(vid).start_tt = tt 
        g.node(vid).end_tt = tt + dtt
        tt += phyllochron
        tiller_points.append((vid, tt + tiller_delay*phyllochron))
    
    for i in range(nb_tillers):
        # add a tiller
        vid, time = tiller_points.pop(0)
    
        tillers = []
        axis_id = g.complex(vid)
        r = g.Rank(vid) + 1  # Number of edges from the root of the axis
        rank_tillers.append(r)
        n = len(g.Axis(vid))
        len_tiller = n - r - 1  # we remove the parent that does not belong to the tiller 
        nb_phy = len_tiller
    
        ranks = range(1, nb_phy + 1)
        ntop = max(ranks) - np.array(ranks) + 1
    
        tt = time
    
        tid = g.add_child(parent=axis_id, edge_type='+', label='Axis')
    
        first = True
        for rank in ranks:
            if first:
                vid, tid2 = g.add_child_and_complex(parent=vid, complex=tid, edge_type='+')
                first = False
            else:
                vid = g.add_child(parent=vid, edge_type='<')
            g.node(vid).start_tt = tt
            g.node(vid).end_tt = tt + dtt
            tt += phyllochron
            tillers.append((vid, tt + tiller_delay*phyllochron))
    
        tiller_points.extend(tillers)
        tiller_points.sort(key=lambda x: x[1]) # sorted by time
        tiller_points = tiller_points[:nb_tillers-i]

    axes = g.vertices(scale=2)  # scale=2: axes
    T = {i+1: [g.order(axis), g.order(axis)+r] for i,(axis,r) in enumerate(zip(axes[1:],rank_tillers))} # {id: (order, delay)}

    return g, T 
```

```python
def resolve_organ_growth(N, nb_tillers, duration, la_ends, tillers, reduction_factor):
    '''
    A: matrix of the advancement of elongation of phytomers i at the end of elongation of phytomers j on the main stem
    T: dict of matrices of the advancement of elongation of phytomers i at the end of elongation of phytomers j on tillers
    B: vector of constrained plant-scale dynamics evaluated at end of elongation of phytomers of the main stem
    Resolve (A+∑T)S=B
    '''
    
    A = np.zeros((N, N)) 
    B = np.zeros(N)
    if nb_tillers > 0:
        T = {i: np.zeros((N, N)) for i in range(1,nb_tillers+1)}
    else:
        T = {}

    for x in range(0, N):
        A[x, min(x+1,N-1)] = (duration - 1) / duration
        for i in range(x+1):
            A[x, i] = 1
        B[x] = la_ends[x] 
        
        if nb_tillers > 0:
            for id,t in T.items():
                delay = tillers[id][1]
                if x >= delay:
                    t[x, min(x+1-delay,N-1)] = (duration - 1) / duration
                    for j in range(x+1): # range(delay+1, x+1) for other hypothesis
                        t[x, max(0,j-delay)] = 1

    # Sum matrices of main stem and tillers 
    P = A
    if nb_tillers > 0:
        for id,t in T.items():
            t *= reduction_factor**tillers[id][0]
            P += t

    # S = np.linalg.solve(P, B)
    S = nnls(P, B)[0]

    return S, P, A, T
```

```python
def resolve_organ_growth_phyllochron(N, nb_tillers, duration, la_ends, tillers, reduction_factor, starts, phyllochron):
    '''
    A: matrix of the advancement of elongation of phytomers i at the end of elongation of phytomers j on the main stem
    T: dict of matrices of the advancement of elongation of phytomers i at the end of elongation of phytomers j on tillers
    B: vector of constrained plant-scale dynamics evaluated at end of elongation of phytomers of the main stem
    Resolve (A+∑T)S=B
    '''

    A = np.zeros((N, N)) 
    B = np.zeros(N)
    if nb_tillers > 0:
        T = {i: np.zeros((N, N)) for i in range(1,nb_tillers+1)}
    else:
        T = {}

    spl = splrep(thermal_time, leaf_area_plant)
    
    for x in range(0, N):
        A[x, x] = 1 / duration
        for i in range(x):
            A[x, i] = 1
        if x < N-1:
            B[x] = splev(starts[x+1], spl)
        else:
            B[x] = splev(starts[N-1]+phyllochron, spl)
        
        if nb_tillers > 0:
            for id,t in T.items():
                delay = tillers[id][1]
                if x >= delay:
                    t[x, min(x+1-delay,N-1)] = 1 / duration
                    for j in range(x+1): # range(delay+1, x+1) for other hypothesis
                        t[x, max(0,j-delay)] = 1

    # Sum matrices of main stem and tillers 
    P = A
    if nb_tillers > 0:
        for id,t in T.items():
            t *= reduction_factor**tillers[id][0]
            P += t

    # S = np.linalg.solve(P, B)
    S = nnls(P, B)[0]

    return S, P, A, T
```

```python
def compute_LA_at_end_of_dev_of_leaves(nb_phy, phyllochron, duration, thermal_time, leaf_area_plant):
    '''
    Evaluate constrained plant-scale dynamics at end of elongation of phytomers of the main stem.
    '''
    starts = []
    ends = []
    for i in range(nb_phy):
        start = i * phyllochron + thermal_time[0]
        end = start + phyllochron * duration
        starts.append(round(start))
        ends.append(round(end))
    spl = splrep(thermal_time, leaf_area_plant)
    la_ends = splev(ends, spl)
    return la_ends, starts, ends
```

```python
from itertools import zip_longest

def sum_lists_right(listes):
    listes_inv = [list(reversed(l)) for l in listes]
    somme_inv = [sum(x) for x in zip_longest(*listes_inv, fillvalue=0)]
    return list(reversed(somme_inv))

sum_lists_right([[1,2,3], [4,5], [6]])
```

```python
def compute_leaf_areas_tillers(S, tillers, reduction_factor, starts, ends):
    S_t = {}
    for id,t in tillers.items():
        order = t[0]
        delay = t[1]
        S_t[id] = [reduction_factor**order*S[i+delay] for i in range(len(S)-delay)]

    S_tot_temp = list(S_t.values())
    S_tot_temp.append(list(S))
    S_tot = sum_lists_right(S_tot_temp)

    S_cum = []
    for t in range(max(ends)):
        sum_temp = 0
        for i,(s,e) in enumerate(zip(starts,ends)):
            if s <= t < e:
                sum_temp += (t-s)/(e-s) * S_tot[i]
            elif t >= e:
                sum_temp += S_tot[i]
        S_cum.append(sum_temp)

    return S_t, S_cum
```

```python
def f(nb_phy = 10, duration = 1.6, nb_tillers = 2, tiller_delay = 1, reduction_factor = 0.8, thermal_time=thermal_time, leaf_area_plant=leaf_area_plant):
    '''
    Plot dynamics of leaf area growth of main stem and tillers, and distribution of leaf areas among leaves of main stem and tillers.
    '''
    phyllochron = (end_veg-thermal_time[0])/(nb_phy + duration)
    print("Phyllochron:", phyllochron)
    g, tillers = generate_tillers(nb_phy, phyllochron, duration, nb_tillers, tiller_delay)
    # print(tillers)
    la_ends, starts, ends = compute_LA_at_end_of_dev_of_leaves(nb_phy, phyllochron, duration, thermal_time, leaf_area_plant)
    S, P, A, T = resolve_organ_growth(nb_phy, nb_tillers, duration, la_ends, tillers, reduction_factor)
    # S, P, A, T = resolve_organ_growth_phyllochron(nb_phy, nb_tillers, duration, la_ends, tillers, reduction_factor, starts, phyllochron)
    S_single, _, _, _ = resolve_organ_growth(nb_phy, 0, duration, la_ends, tillers, reduction_factor)
    S_t, S_cum = compute_leaf_areas_tillers(S, tillers, reduction_factor, starts, ends)

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8), width_ratios=[2,1])
    ax1.plot(thermal_time, leaf_area_plant, linewidth=2, color="black", label="Crop model")
    for i in range(len(S)):
        if i == 1:
            ax1.plot(np.linspace(starts[i], ends[i], 2), [0, S[i]], color="darkorange", linewidth=2, label="Area of each leaf")
            for id,s in S_t.items():
                delay = tillers[id][1]
                if i+delay < len(S):
                    ax1.plot(np.linspace(starts[i+delay], ends[i+delay], 2), [0, s[i]], linewidth=2, color="darkorange")
        elif i > 1:
            ax1.plot(np.linspace(starts[i], ends[i], 2), [sum(S[:i]), sum(S[:i+1])], linewidth=2, color="darkorange")
            for id,s in S_t.items():
                delay = tillers[id][1]
                if i+delay < len(S):
                    ax1.plot(np.linspace(starts[i+delay], ends[i+delay], 2), [sum(s[:i]), sum(s[:i+1])], linewidth=2, color="darkorange")

    ax1.fill_between(range(len(S_cum)), [0]*len(S_cum), S_cum, color="orange", alpha=0.15, label="Cumulated sum of leaf areas")
    ax2.scatter(S, range(1,nb_phy+1), color='darkorange', label='Main stem and tillers')
    ax2.plot(S, range(1,nb_phy+1), color='darkorange')
    for id,s_t in S_t.items():
        delay = tillers[id][1]
        ax2.scatter(s_t, range(1+delay,nb_phy+1), color='darkorange')
        ax2.plot(s_t, range(1+delay,nb_phy+1), color='darkorange')
    ax2.scatter(S_single, range(1,nb_phy+1), color='red', label='Single stem')
    ax2.plot(S_single, range(1,nb_phy+1), color='red')
    ax2.plot([0]*nb_phy, range(1,nb_phy+1), color='grey', linestyle='--')
    ax1.legend()
    ax2.legend()
    plt.show()
```
