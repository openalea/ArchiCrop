```python
import os
from datetime import date

from openalea.widgets.plantgl import *
from openalea.plantgl.all import Material, Color3, Scene, Viewer
from openalea.archicrop.export_mtg import save_mtg
from pathlib import Path

import sys
sys.path.append('../data')
from archi_dict import *

from openalea.archicrop.archicrop import ArchiCrop
from openalea.archicrop.display import build_scene, display_scene
from openalea.archicrop.stand import agronomic_plot, compute_domain, regular, regularband, config_crop_intercrop
```

```python
green = Color3((0,50,0))
conv_coef = 100
```

```python
dates = [0]
```

```python
cereals = ["Sorghum", "Maize", "Wheat", "Rice 1", "Rice 2"]

cereal_dict = {}

for c, archi in zip(cereals, [archi_sorghum_theo, archi_maize_theo, archi_wheat_theo, archi_rice_theo_1, archi_rice_theo_2]):
    cereal = ArchiCrop(**archi, plant_orientation=90)
    cereal.generate_potential_plant()
    cereal_dict[c] = cereal.g

sorghum = cereal_dict["Sorghum"]
maize = cereal_dict["Maize"]
wheat = cereal_dict["Wheat"]
rice_1 = cereal_dict["Rice 1"]
rice_2 = cereal_dict["Rice 2"]
```

```python
crop_1 = {0:maize}
crop_2 = {0:wheat}
```

```python
design = "strips"

density_1 = 8
density_2 = 50
inter_row_1 = 0.70
inter_row_2 = 0.20
nb_rows_1 = 2
nb_rows_2 = 5
# width = 8
# length = 2
width = 2 * inter_row_1 if design == "mixed" else (nb_rows_1-1) * inter_row_1 + (nb_rows_2-1) * inter_row_2 + 2*max(inter_row_1, inter_row_2)
length = 2 * max(1/density_1/inter_row_1, 1/density_2/inter_row_2) if design == "mixed" else max(1/density_1/inter_row_1, 1/density_2/inter_row_2)

list_of_graphs, positions_crop = config_crop_intercrop(dates, crop_1, crop_2, inter_row_1, inter_row_2, density_1, density_2, nb_rows_1, nb_rows_2, width, length, "N-S")
scene, labels = build_scene(list_of_graphs[0], positions_crop, leaf_material=Material(green), stem_material=Material(green), senescence=True)
Viewer.display(scene)
```

```python
for k,g in enumerate(list_of_graphs[0]):
    scene, labels = build_scene(mtg=g, position=positions_crop[k], senescence=True)
    mtg_fn = Path(f"../simulations_ArchiCrop/{k}.mtg")
    obj_fn = Path(f"../simulations_ArchiCrop/{k}.obj")
    save_mtg(g, scene, mtg_fn, obj_fn)
```
